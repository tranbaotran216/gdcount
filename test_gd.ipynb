```python

from models.gdcount_model import GroundingDINOCounter, GDCountConfig
from gdcount.data_utils import preprocess_inference_image
import torch
from torch.utils.data import DataLoader
from datasets.fsc147_dataset import FSC147Dataset, fsc147_collate

device = "cuda" if torch.cuda.is_available() else "cpu"

train_ds = FSC147Dataset(
    ann_path="data/FSC147/annotation_FSC147_384.json",
    img_root="data/images_384_VarV2",                 # hoặc đường dẫn tuyệt đối tới images_384_VarV2
    split="train",
    split_file="data/FSC147/Train_Test_Val_FSC_147.json",
    class_map_file="data/FSC147/ImageClasses_FSC147.txt",
    img_size=None,                                    # nếu dataset của bạn đã resize sẵn 384
    normalize=True,
    density_root=None,
)

train_loader = DataLoader(
    train_ds,
    batch_size=1,
    shuffle=True,
    num_workers=0,
    collate_fn=fsc147_collate,
)

batch = next(iter(train_loader))
print(batch.keys())
print(batch["images"].shape, batch["gt_counts"], batch["prompts"][0])

```
